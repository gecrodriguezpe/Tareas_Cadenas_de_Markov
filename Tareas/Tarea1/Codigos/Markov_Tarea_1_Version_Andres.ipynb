```python
from random import uniform

p_c1 = 0.19634954
lst_p_delta_c1 = []

print("1.\n")
print(f"n\tDC\tP delta\tError abs")

for i in range(2, 7):
    frec_abs = 0
    for j in range(10**i):
        x = uniform(-2, 2)
        y = uniform(-2, 2)
        if x*x + y*y <= 1:
            frec_abs += 1

    p_delta_c1 = frec_abs / 10**i
    lst_p_delta_c1.append(p_delta_c1)
    error_abs = abs(p_c1 - p_delta_c1)

    print(f"10^{i}\t{frec_abs}\t{p_delta_c1}\t{error_abs}")
```

```python
print("\n2.\n")

print(f"n\tP delta\tPi aprox\tError abs pi")

for i in range(2, 7):
    p_delta_c1_i = lst_p_delta_c1[i - 2]
    pi_aprox = 16 * p_delta_c1_i
    error_abs_pi = abs(3.14159265 - pi_aprox)
    print(f"10^{i}\t{p_delta_c1_i}\t{pi_aprox}\t{error_abs_pi}")
```

```python
print("\n3.\n") #Recordar Area del circulo / Area del cuadrado

print(f"i\tP delta\tPi aprox\tError abs pi")
for k in range(1, 3):
    m = 2 * k
    for i in range(2, 7):
        frec_abs = 0
        for j in range(10**i):
            x = uniform(-2, 2)
            y = uniform(-2, 2)
            if x*x + y*y <= m:
                frec_abs += 1

        p_delta_c1 = frec_abs / 10**i
        pi_aprox = (16 / m) * p_delta_c1
        error_abs_pi = abs(3.14159265 - pi_aprox)

        print(f"10^{i}\t{p_delta_c1}\t{pi_aprox}\t{error_abs_pi}")
```
